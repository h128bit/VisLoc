```python
from geopy.distance import distance
import numpy as np
from pyproj import Transformer
import numpy as np
from PIL import Image, ImageDraw
import io 
import holoviews as hv

lat=-17.120500564575195
lon=-71.79552459716797
start = (lat, lon)

north = distance(meters=75).destination(start, bearing=0).latitude    # На север
south = distance(meters=75).destination(start, bearing=180).latitude  # На юг
east  = distance(meters=75).destination(start, bearing=90).longitude   # На восток
west  = distance(meters=75).destination(start, bearing=270).longitude  # На запад

bbox = (west, south, east, north)
```

```python
from mapminer.miners import GoogleBaseMapMiner
miner = GoogleBaseMapMiner()
# lat=-17.120500564575195 
# lon=-71.79552459716797

# lat, lon = 37.288455963134766, -120.36018371582031
# lat, lon = 64.10420227050781, -21.831266403198242
lat, lon = 54.77445983886719, 32.0266227722168
# lat, lon = 54.82347869873047, 56.080379486083984
t = Transformer.from_crs("EPSG:4326", "EPSG:3857", always_xy=True)
x, y = t.transform(lon, lat)

t = Transformer.from_crs("EPSG:3857", "EPSG:4326", always_xy=True)
lon_t, lat_t = t.transform(x+0, y-0)

ds1 = miner.fetch(lat=lat_t, lon=lon_t, radius=80, resolution=2)
```

```python
## For GoogleBaseMapMiner
t = Transformer.from_crs("EPSG:4326", "EPSG:3857", always_xy=True)
x, y = t.transform(lon, lat)

x_range = ds1["x"].to_numpy()
y_range = ds1["y"].to_numpy()

idx_x = np.abs(x_range - x).argmin()
idx_y = np.abs(y_range - y).argmin()
print(idx_x, idx_y)

plot = ds1.to_numpy()
plot = np.transpose(plot, (1, 2, 0))
shape = plot.shape
print(shape)
img = Image.fromarray(plot)
img1 = img.copy()
draw = ImageDraw.Draw(img1)
draw.circle([idx_x, idx_y], radius=5, fill="red",)
img1
```
